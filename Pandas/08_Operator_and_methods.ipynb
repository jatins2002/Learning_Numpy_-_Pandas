```python
import numpy as np
import pandas as pd
```

```python
salary = np.array([45000, 52000, 0, 61000, 48000])
department = np.array(["developer", "marketing", "intern", "sales", "support"])

salary_series = pd.Series(salary, index=department, name="Salary")
salary_series
```

### Arithmetic Operator & Methods

```python
#Addition
#salary_series + 2
salary_series.add(2)
```

```python
#Subtraction
#salary_series - 2
salary_series.sub(2)
```

```python
#Multiplication
#salary_series * 2
salary_series.mul(2)
```

```python
#Division
#salary_series / 2
salary_series.div(2)
```

```python
#subtraction
#salary_series // 2
salary_series.floordiv(2)
```

```python
#subtraction
#salary_series % 2
salary_series.mod(2)
```

```python
#subtraction
#salary_series ** 2
salary_series.pow(2)
```
