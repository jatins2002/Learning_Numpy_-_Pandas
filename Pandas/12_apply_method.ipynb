```python
import pandas as pd

billing = pd.Series([450.75, 1200.50, 980.00,
                     1550.25, 760.40, 2300.00,
                     890.10, 1475.60, 650.00,
                     3100.75], name="billing")
billing
```

```python
def give_discount(price : float):
    if price > 1000:
        price = price - (price * 0.1)
        return round(price, 2)
    return price
```

```python
billing.apply(give_discount)
```

```python


students = ["Aarav", "Diya", "Kabir", "Meera", "Rohan", "Sana", "Vikram", "Zoya"]

marks = pd.Series([92, 78, 35, 66, 88, 49, 57, 95], index=students, name="marks")
marks
```

```python
def get_grade(score:int):
    if score >= 90:
        return "A"
    elif score >= 75:
        return "B"
    elif score >= 60:
        return "C"
    elif score >= 40:
        return "D"
    else:
        return "F"
```

```python
grades = marks.apply(get_grade)
grades
```

```python
grades.value_counts()
```
