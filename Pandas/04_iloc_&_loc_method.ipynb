```python
import pandas as pd
```

### loc Methods

```python
salary = [45000,57000,0,61000,58000,72000]
dept = ["developer","marketing","hr","sales","finance","support"]

salary_series = pd.Series(salary, index=dept, name = "salary")
```

```python
salary_series.loc["developer"]
```

```python
salary_series.loc["hr"]
```

```python
salary_series.loc[["hr","marketing"]]
```

```python
salary_series.loc["marketing":"support"] #loc includes support not other python list that exclude the stop value.
```

```python
salary_series.loc["marketing":"support":2]
```

### iloc methods

```python
salary = [45000,57000,0,61000,58000,72000]
dept = ["developer","marketing","hr","sales","finance","support"]

salary_series = pd.Series(salary, index=dept, name = "salary")
```

```python
salary_series.loc["developer"]
# OR
salary_series.iloc[0]
```

```python
salary_series.loc["hr"]
# OR
salary_series.iloc[2]
```

```python
salary_series.loc[["hr","marketing"]]
# OR
salary_series.iloc[[2,1]]
```

```python
salary_series.loc["marketing":"support"]
# OR
salary_series.iloc[1:5]
```

```python
#salary_series.loc["marketing":"support":2]
# OR
salary_series.iloc[1:5:2] #in iloc excludes support  other python list that exclude the stop value.
```
