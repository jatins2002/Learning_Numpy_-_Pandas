```python
import pandas as pd
```

```python
salary = [45000,57000,0,61000,58000,72000]
dept = ["developer","marketing","hr","sales","finance","support"]

salary_series = pd.Series(salary, index=dept, name = "salary")
```

### Indexing

```python
salary_series["developer"]
```

```python
salary_series["marketing"]
```

```python
salary_series[["sales","support"]]
```
