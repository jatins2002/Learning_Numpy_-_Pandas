```python
battery = np.array([20,45,8,6,13])
is_charging = np.array(["True","False","False","True","False"])
#     "Charging"   -> if the phone is plugged in
#     "Charge Now" -> if NOT plugged in AND battery < 20
#     "OK"         -> otherwise

status = np.where(is_charging == "True","Charging",np.where(battery < 20,"Charge Now","OK"))
status
```

```python
#Stock report using yesterday's and today's prices:
#     First find the % change = (today - yesterday) / yesterday * 100
#     Then label: "Big Jump" if change > 5, "Up" if change > 0,
#                 "Down" if change < 0, else "No Change"
#     Hint: put one np.where inside another
yesterday = np.array([200, 150, 320, 90, 500, 75])
today     = np.array([215, 150, 310, 93, 500, 72])
# Your code here
change = (today - yesterday) / yesterday * 100
change
```

```python
label = np.where(change > 5, "Big Jump",
                 np.where(change > 0, "Up",
                          np.where(change < 0, "Down", "No Change")))
label
```
