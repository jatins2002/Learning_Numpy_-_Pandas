```python
import numpy as np
```

```python
ages = np.array([18, 22, 25, 30, 21, 28, 35, 40, 19, 24, 32, 27, 45, 23, 29])
```

```python
#change one value
ages[0] = 20
ages
```

```python
#Replace a slice
ages[1:3] = [70,75]
ages
```

```python
#update with condition
ages[ages < 20] = 20
ages
```

```python
#update selected points
ages[[3,4]] = [60,90]
ages
```
