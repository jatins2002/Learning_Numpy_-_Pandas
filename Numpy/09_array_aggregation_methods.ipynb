```python
import numpy as np
```

### 1-D Array

```python
sales = np.array([120, 200, 150, 180, 210, 190, 170])
```

```python
#total sales
sales.sum()
```

```python
#sales average
np.round(sales.mean(),2)
```

```python
#High sales
sales.max()
```

```python
#lowest sales
sales.min()
```

### 2-D Array

```python
# axis=0 -> down the rows, one result per column | axis=1 -> across columns, one result per row
sales = np.array([[120, 150, 180, 200],
                  [100, 170, 160, 210],
                  [140, 130, 190, 220]])
```

```python
# whole array total
sales.sum()
```

```python
# column-wise sum
sales.sum(axis=0)
```

```python
# Row-wise sum
sales.sum(axis = 1)
```

```python
# Column-wise average
sales.mean(axis = 0)
```

```python
# Row-wise average
sales.mean(axis = 1)
```

```python
#MEDIAN

np.median(sales)
```

```python
#PERCENTILE

np.percentile(sales, 75)
```

```python
sales = np.array([150, 200, 150, 300, 250, 200, 350, 150])  # has duplicates like 150 and 200
```

```python
# Unique values: distinct values, sorted
np.unique(sales)

# Unique values AND how many times each appears
np.unique(sales, return_counts=True)
```

```python
# Square Root
np.sqrt(sales)
```

```python
# ==============================
# MORE USEFUL FUNCTIONS
# ==============================
sales = np.array([120, 200, 150, 180, 210, 190, 170])
```

```python
# Standard deviation: how spread out the values are
sales.std()
```

```python
# Position (index) of the highest and lowest sale
print(sales.argmax(), sales.argmin())
```

```python
# Running total: each value = sum of all values up to that point
sales.cumsum()
```

```python
# Sort values from smallest to largest
# sales.sort()
sorted_sales = np.sort(sales)
print(sorted_sales)
print(sales)
```

```python
# Absolute value: removes the minus sign (e.g. profit/loss -> size of change)
changes = np.array([-20, 15, -5, 30, -10])
np.abs(changes)
```
