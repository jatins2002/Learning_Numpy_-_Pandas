```python
import numpy as np
```

```python
ages = np.array([15, 2, 5, 30, 21, 14, 3, 40, 19, 24, 32, 27, 4, 23, 9])
```

```python
adult_mask = ages >= 18
adult_mask
```

```python
ages[adult_mask]
```

```python
ages[ages >= 18]
```

```python
adult = ages[ages >= 18]
adult
```

```python
child_mask = ages < 18
child_mask
```

```python
ages[child_mask]
```

```python
child = ages[ages < 18]
child
```

```python
teen_mask = (ages <= 15) & (ages >= 10)
ages[teen_mask]
```

```python
special_mask = (ages > 10) | (ages < 60)
ages[special_mask]
```

### Filtering One array using Another Array

```python
ages = np.array([18, 22, 25, 30, 21, 28, 35, 40, 19, 24, 32, 27, 45, 23, 29])

person = np.array([
    "Rahul", "Priya", "Amit", "Sneha", "Vikash",
    "Neha", "Rohit", "Anjali", "Karan", "Pooja",
    "Arjun", "Simran", "Aditya", "Riya", "Varun"
])
```

```python
adult_mask = ages >= 18
ages[adult_mask]
person[adult_mask]
```
